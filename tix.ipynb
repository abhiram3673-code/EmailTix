```python
df['Product Purchased']=df['Product Purchased'].fillna('device')
df['Ticket Description']=df['Ticket Description'].fillna('')
def merge_p_t(row):
    desc=row['Ticket Description']
    prod=str(row['Product Purchased'])
    desc=desc.replace('{product_purchased}',prod)
    return desc
df['Ticket Description']=df.apply(merge_p_t,axis=1)
```
