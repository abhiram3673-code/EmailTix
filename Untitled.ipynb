```python

import re

df = pd.read_csv('customer_support_tickets.csv')

# 1. Fill missing text and product values
df['Ticket Subject'] = df['Ticket Subject'].fillna('')
df['Ticket Description'] = df['Ticket Description'].fillna('')
df['Product Purchased'] = df['Product Purchased'].fillna('device')

# 2. Dynamically replace '{product_purchased}' with the actual product name (or 'device')
def clean_description(row):
    desc = row['Ticket Description']
    product = str(row['Product Purchased']).strip()
    # Replace literal template string '{product_purchased}' with the actual item name
    clean_desc = desc.replace('{product_purchased}', product)
    # Remove leftover template noise/artifacts if any remain
    clean_desc = re.sub(r'\{.*?\}', product, clean_desc)
    return clean_desc

df['clean_description'] = df.apply(clean_description, axis=1)

# 3. Combine Subject + Cleaned Description into a single NLP feature column
df['full_feature_text'] = df['Ticket Subject'] + " " + df['clean_description']

# 4. Now safe to drop metadata columns!
df_final = df[['full_feature_text', 'Ticket Type', 'Ticket Priority']].dropna()
```
